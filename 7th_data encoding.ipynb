```python
import pandas as pd
import numpy as np
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({'status' : ['single', 'married', 'separated','single', 'married', 'separated','single', 'married', 'separated','single', 'married', 'separated','single', 'married', 'separated','single', 'married', 'separated','single', 'married', 'separated','single', 'married', 'separated','single', 'married', 'separated','single', 'married', 'separated']})
```

```python
df
```

```python
encoder = OneHotEncoder()
```

```python
encoder
```

```python
encoded = encoder.fit_transform(df[['status']]).toarray()
```

```python
encoder.get_feature_names_out()
```

```python
encoder_df = pd.DataFrame(encoded, columns= encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
#new data 
encoder.transform([['single']]).toarray()
```

```python
pd.concat([df,encoder_df],axis=1)
```

```python
df_dummies = pd.get_dummies(df, columns=['status'])
```

```python
#TARGET GUIDED ORDINAL ENCODING
#relationship with garget variable
#a lot of unique catagories 
# replace the category with the mean and median of respective groupes target variable
```

```python
df = pd.DataFrame({'time': ['lunch', 'breakfirst', 'dinner','lunch', 'breakfirst', 'dinner','lunch', 'breakfirst', 'dinner'],
                   'total_bill': [120, 130, 90, 125, 150, 190, 160, 198, 45]})
```

```python
df
```

```python
mean_price = df.groupby('time')['total_bill'].mean().to_dict()
```

```python
mean_price
```

```python
df['time_encoded'] = df['time'].map(mean_price)
```

```python
df
```
